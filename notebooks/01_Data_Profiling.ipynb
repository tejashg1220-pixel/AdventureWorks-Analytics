```python
sales_2016 = pd.read_csv(
    raw_path / "Sales-2016.csv"
)

sales_2017 = pd.read_csv(
    raw_path / "Sales-2017.csv"
)

for year, df in [
    ("2015", sales_2015),
    ("2016", sales_2016),
    ("2017", sales_2017)
]:
    order_date = pd.to_datetime(df["OrderDate"])
    stock_date = pd.to_datetime(df["StockDate"])

    difference = (stock_date - order_date).dt.days

    print(f"\n--- {year} ---")
    print("Rows:", len(df))
    print("OrderDate range:",
          order_date.min().date(), "to", order_date.max().date())
    print("StockDate range:",
          stock_date.min().date(), "to", stock_date.max().date())
    print("Minimum difference:", difference.min())
    print("Maximum difference:", difference.max())
    print("StockDate after OrderDate:",
          (difference > 0).sum())
```

```python
# RFM Analysis - Step 1
# Create the base RFM table

import pandas as pd

# Load analysis-ready sales data
sales_rfm = pd.read_csv(
    "../data/cleaned/Sales_Analysis_Ready.csv"
)

# Convert OrderDate to datetime
sales_rfm["OrderDate"] = pd.to_datetime(
    sales_rfm["OrderDate"],
    format="%Y-%m-%d"
)

# Use the latest transaction date as the analysis reference date
reference_date = sales_rfm["OrderDate"].max()

print("Reference Date:", reference_date)

# Create customer-level RFM metrics
rfm = sales_rfm.groupby("CustomerKey").agg(
    Recency=("OrderDate", lambda x: (reference_date - x.max()).days),
    Frequency=("OrderNumber", "nunique"),
    Monetary=("Revenue", "sum")
).reset_index()

# Display first few records
rfm.head()
```

```python
# RFM Analysis - Step 2
# Create Recency, Frequency and Monetary scores

# Recency:
# Lower recency is better, so the scoring is reversed.
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency:
# Higher frequency is better.
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary:
# Higher monetary value is better.
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Combined RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"]
    + rfm["F_Score"]
    + rfm["M_Score"]
)

rfm.head()
```

```python
# RFM Analysis - Step 3
# Create customer segments

def assign_rfm_segment(row):
    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    elif r >= 3 and f >= 4 and m >= 3:
        return "Loyal Customers"

    elif r >= 4 and f <= 2 and m <= 3:
        return "New Customers"

    elif r <= 2 and f >= 3 and m >= 3:
        return "At Risk"

    elif r <= 2 and f <= 2 and m >= 3:
        return "Can't Lose Them"

    elif r <= 2 and f <= 2 and m <= 2:
        return "Lost Customers"

    else:
        return "Potential Loyalists"


rfm["Customer_Segment"] = rfm.apply(
    assign_rfm_segment,
    axis=1
)

# Check segment distribution
rfm["Customer_Segment"].value_counts()
```

```python
# RFM Analysis - Step 4
# Create segment-level business summary

rfm_summary = (
    rfm.groupby("Customer_Segment")
    .agg(
        Customers=("CustomerKey", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Total_Revenue=("Monetary", "sum"),
        Avg_Revenue_Per_Customer=("Monetary", "mean")
    )
    .reset_index()
)

# Add revenue share
rfm_summary["Revenue_Share_%"] = (
    rfm_summary["Total_Revenue"]
    / rfm_summary["Total_Revenue"].sum()
    * 100
)

# Round values for readability
rfm_summary["Avg_Recency"] = rfm_summary["Avg_Recency"].round(1)
rfm_summary["Avg_Frequency"] = rfm_summary["Avg_Frequency"].round(2)
rfm_summary["Total_Revenue"] = rfm_summary["Total_Revenue"].round(2)
rfm_summary["Avg_Revenue_Per_Customer"] = (
    rfm_summary["Avg_Revenue_Per_Customer"].round(2)
)
rfm_summary["Revenue_Share_%"] = (
    rfm_summary["Revenue_Share_%"].round(2)
)

# Sort by revenue
rfm_summary = rfm_summary.sort_values(
    "Total_Revenue",
    ascending=False
)

rfm_summary
```

```python
# RFM Analysis - Step 5
# Save customer-level RFM dataset

rfm.to_csv(
    "../data/cleaned/RFM_Customer_Segmentation.csv",
    index=False
)

print("RFM dataset saved successfully.")
print("Shape:", rfm.shape)
```
